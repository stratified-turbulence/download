- This notebook allows you to download specified subvolumes through x,y,z ranges from any sparsing level, variable, simulation

**Instructions:**

Run the following cells in order:
1. Configure (specify required settings and connect to Constellation repository)
2. Download specified data (either live through HTTPS or through generated terminal commands via Globus CLI)
3. Visualize downloaded datasets (can run this without step 2 if you've previously saved data)

## 1 · Configure
Specify requested parameters, then run the cell to connect to the repository:

```python
import zarr_app

# EDIT PARAMETERS IN THIS CELL
CONFIG = {
    "CACHE_DIR":    "/absolute/path/to/cache/dir",   # Specify path to folder where intermediate downloaded chunks can be stored here (safe to clear once data download is complete).
                              # If using Globus Connect option in Step 2, must allow Globus access to write in this directory.
    "SAVE_DIR":     "/absolute/path/to/save/dir",   # Specify path to folder where final reconstructed .npy data will land 
    "MAX_FETCH_GB": 500,                 # Refuse selections larger than this (to prevent accidental disk filling)
    "DL_WORKERS":   12,                  # Number of concurrent chunk downloads (1 = serial, 2+ provides speed up, test based on your machine)
    "DEST_ENDPOINT": "xxxx",  # (Optional) If using Globus connect endpoint, enter your Endpoint ID (see Appendix A, readme)
}

zarr_app.configure(CONFIG)
```

## 2 · Download

Pick simulation, variable, sparsing level, and x, y, z index ranges. 

Sliders snap to the inner-chunk grid (smallest downloadable unit).

Transfer estimate shows size of data that must be downloaded to reconstruct subvolume. 

The result saves to the SAVE_DIR folder specified in Configure cell.

**For small jobs:** Download live via HTTPS, download and reconstruct in real time through this notebook.

**For larger jobs:** If you have the Globus app on your local machine, copy and run generated terminal command on local machine to use Globus transfer (handles restarts and checkpointing automatically). This can be done outside of this notebook. This notebook can be used to generate commands to obtain multiple desired subvolumes, which are then downloaded outside of the notebook. **If using this option, must first run globus login in your terminal.**

To stop a job, inturrupt the cell execution and rerun cell.

```python
zarr_app.download_panel()
```

## 3 · Visualize

Load using memory maps and visualize any saved `.npy` (located in the SAVE_DIR folder from Configure cell).

Press **Refresh** after downloading a new cube.

Visualizing slices with side lengths longer than 2000 pixels may be relatively slow, depending on your computer.

```python
zarr_app.visualize_panel()
```
